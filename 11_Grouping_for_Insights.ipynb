#Import data from kaggle


```python
import pandas as pd
dd = pd.read_csv("Sales_December_2019.csv")
df = pd.read_csv("Sales_February_2019.csv")
dd.head()
```

### Process and Aggregate Sales Data

```python
all_data = pd.concat([dd, df], ignore_index=True)

# Convert 'Order Date' to datetime objects
all_data['Order Date'] = pd.to_datetime(all_data['Order Date'], errors='coerce')

# Drop rows with NaN values that resulted from coercion
all_data.dropna(subset=['Order Date', 'Quantity Ordered', 'Price Each'], inplace=True)

# Convert 'Quantity Ordered' and 'Price Each' to numeric types
all_data['Quantity Ordered'] = pd.to_numeric(all_data['Quantity Ordered'])
all_data['Price Each'] = pd.to_numeric(all_data['Price Each'])

# Extract month and year
all_data['Month'] = all_data['Order Date'].dt.month
all_data['Year'] = all_data['Order Date'].dt.year
```

### Multi-level Grouping and Aggregation with `groupby()` and `.agg()`

```python
# Multi-level grouping and aggregation
monthly_sales = all_data.groupby(['Year', 'Month']).agg(
    Total_Quantity=('Quantity Ordered', 'sum'),
    Total_Sales=('Price Each', lambda x: (x * all_data.loc[x.index, 'Quantity Ordered']).sum()),
    Number_of_Orders=('Order ID', 'count')
).reset_index()

display(monthly_sales)
```

#Load the dataset from kaggel API

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ibrohimsharipov/12-months-sales-data")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

# List all CSV files in the downloaded directory
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]

# Create an empty list to store dataframes
dfs = []

# Read each CSV file and append it to the list of dataframes
for csv_file in csv_files:
    file_path = os.path.join(path, csv_file)
    df = pd.read_csv(file_path)
    dfs.append(df)

# Concatenate all dataframes into a single dataframe
df = pd.concat(dfs, ignore_index=True)

# Display the first few rows of the DataFrame
display(df.head())
```

#New Column Creation
Refers to adding one or more new columns to an existing DataFrame. This is done to derive new information from existing data or to prepare the data for further analysis.

```python
# Convert 'Order Date' to datetime objects
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

# Extract month and hour into new columns
df['Order Month'] = df['Order Date'].dt.month
df['Order Hour'] = df['Order Date'].dt.hour

# Display the first few rows with the new columns
display(df.head())
```

#Column Arithmetic
Refers to performing mathematical operations on one or more columns within a DataFrame to create new values or derive insights.

```python
# Calculate total sales for each month
monthly_sales = df.groupby('Order Month')['Total Price'].sum()

# Display the monthly sales
display(monthly_sales)
```

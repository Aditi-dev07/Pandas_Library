#Load the Dataset

```python
import pandas as pd
df= pd.read_csv("Datafiniti_Womens_Shoes.csv")
df.head()
```

#sort_values() is a function in the pandas library that allows you to sort a DataFrame by the values in one or more columns. You can specify whether to sort in ascending or descending order

```python
df_sorted = df.sort_values(by='dateUpdated', ascending=False)
display(df_sorted.head())
```

```python
# Sort by a single column in ascending order
df_sorted_asc = df.sort_values(by='dateAdded', ascending=False)
print("Sorting by 'dateAdded' in ascending order:")
display(df_sorted_asc.head())

# Sort by multiple columns
# Sort by 'brand' in ascending order and then by 'dateAdded' in descending order
df_sorted_multi = df.sort_values(by=['brand', 'dateAdded'], ascending=[False , False])
print("\nSorting by 'brand' (ascending) and 'dateAdded' (descending):")
display(df_sorted_multi.head())
```

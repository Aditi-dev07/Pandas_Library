Load the dataset
https://www.kaggle.com/datasets/ibrohimsharipov/12-months-sales-data#Import

```python
import pandas as pd
dfd = pd.read_csv("Sales_December_2019.csv")
dff = pd.read_csv("Sales_February_2019.csv")
```

The .head() function in pandas is used to display the first n rows of a DataFrame or Series

```python
all_data = pd.concat([dfd, dff])
display(all_data.head())
```

## Identify multi-item orders

### Subtask:
Create a new dataframe containing only the rows corresponding to duplicated 'Order ID' values.


**Reasoning**:
Identify and filter for rows with duplicated 'Order ID' values, including the first occurrence, and display the head of the resulting dataframe.



```python
df_duplicated = all_data[all_data['Order ID'].duplicated(keep=False)]
display(df_duplicated.head())
```

## Group and join products

### Subtask:
Group the new dataframe by 'Order ID' and join the 'Product' names into a single string for each order.


**Reasoning**:
Group the DataFrame by 'Order ID' and aggregate the 'Product' column into a single string.



```python
grouped_df = df_duplicated.groupby('Order ID')['Product'].apply(lambda x: ', '.join(x)).reset_index()
display(grouped_df.head())
```

## Drop duplicates

### Subtask:
Drop duplicate rows based on 'Order ID'.


**Reasoning**:
Drop duplicate rows based on 'Order ID' and display the head of the resulting DataFrame.



```python
grouped_df = grouped_df.drop_duplicates(subset=['Order ID'])
display(grouped_df.head())
```

## Count product bundles

### Subtask:
Use `.value_counts()` on the 'Grouped' column to find the most common combinations of products sold together.


**Reasoning**:
Apply value_counts to the 'Product' column of grouped_df to find the frequency of each product bundle and display the top few results.



```python
product_bundle_counts = grouped_df['Product'].value_counts()
display(product_bundle_counts.head())
```

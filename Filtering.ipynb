##Load dataset from kaggel
#import pandas library

```python
import pandas as pd
df= pd.read_csv("data.csv")
df.head()
```

##Boolean Indexing
Boolean indexing means selecting rows from a DataFrame (or elements from a Series) based on a condition that returns True/False values.

```python
referral_channel = df[df["channel"] == "referral"]
print(referral_channel)
```

```python
print("Cost > 10:")
print(df[df["cost"] > 10])
```

##AND operator (both conditions must be true).

```python
print("\nCost > 10 AND Revenue > 3000:")
print(df[(df["cost"] > 10) & (df["revenue"] > 3000)])
```

##OR operator (at least one condition must be true).

```python
print("\nCost < 6 OR Conversion Rate > 0.15:")
print(df[(df["cost"] < 6) | (df["conversion_rate"] > 0.15)])
```

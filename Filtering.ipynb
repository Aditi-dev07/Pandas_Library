##Boolean Indexing
Boolean indexing means selecting rows from a DataFrame (or elements from a Series) based on a condition that returns True/False values.

```python
print("Transactions with amount > 100:")
print(df[df["Transaction Amount"] > 100])
```

```python
print("\nTransactions where Category is NOT 'Electronics':")
print(df[df["Category"] != "Electronics"])
```

##AND operator (both conditions must be true).

```python
print("\nTransactions with Amount > 100 AND Customer ID > 500000:")
print(df[(df["Transaction Amount"] > 100) & (df["Customer ID"] > 500000)])
```

##OR operator (at least one condition must be true).

```python
print("\nTransactions with Amount > 100 OR Category is 'Travel':")
print(df[(df["Transaction Amount"] > 100) | (df["Category"] == "Travel")])
```

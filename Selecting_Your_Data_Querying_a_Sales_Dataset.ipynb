#Load the dataset from Kaggel

```python
import pandas as pd
import os

# Direct download link from Kaggle
download_url = "https://www.kaggle.com/api/v1/datasets/download/ibrohimsharipov/12-months-sales-data?dataset_version_number=1"
output_filename = "12-months-sales-data.zip" # Define an output filename for the downloaded zip

# Download the zip file using wget
!wget -O {output_filename} "{download_url}"

# Unzip the downloaded file
!unzip {output_filename} -d ./sales_data

# Find the actual data file (assuming it's a CSV in the unzipped directory)
# You might need to adjust the filename based on the contents of the zip
extracted_files = os.listdir('./sales_data')
print("Files extracted:", extracted_files)

# Assuming the data is in a CSV file, let's try to load it
# You might need to adjust the filename based on the output of os.listdir()
try:
    # Find the first CSV file in the extracted directory
    csv_files = [f for f in extracted_files if f.endswith('.csv')]
    if csv_files:
        data_file_path = os.path.join('./sales_data', csv_files[0])
        df = pd.read_csv(data_file_path)
        print("\nData loaded successfully!")
        display(df.head())
    else:
        print("\nError: No CSV files found in the extracted directory.")

except FileNotFoundError:
    print("\nError: Data file not found. Please check the filename and path.")
except Exception as e:
    print(f"\nAn error occurred while loading the data: {e}")
```

#**loc stands for location**.

**It is used for label-based indexing.**

**You select rows & columns by their labels (names), not by integer positions.**

**Syntax**

**df.loc[row_selection, column_selection]**

```python
# Select rows with index 0 to 4 (labels) and columns 'Product' and 'Price Each' (labels)
subset_df = df.loc[0:4, ['Product', 'Price Each']]

display(subset_df)

# You can also use it to select rows based on a condition
# For example, to select all rows where 'Quantity Ordered' is greater than 1
# Note: You might need to convert 'Quantity Ordered' to a numeric type first if it's not already
try:
    df['Quantity Ordered'] = pd.to_numeric(df['Quantity Ordered'])
    high_quantity_orders = df.loc[df['Quantity Ordered'] > 1]
    print("\nOrders with Quantity Ordered > 1:")
    display(high_quantity_orders.head())
except ValueError:
    print("\nCould not convert 'Quantity Ordered' to numeric. Please check the data for non-numeric values.")
```

#**iloc in pandas stands for index location.**

####**Is used for integer-based indexing (row/column positions).you select rows and columns by their numeric position (0, 1, 2, …)**

#####**The syntax is:**


**df.iloc[ row_selection , column_selection ]**

```python
# Another example using .iloc (integer-based indexing)
# Let's select rows from the 10th row up to (but not including) the 20th row,
# and the first three columns (at integer positions 0, 1, and 2).
try:
    subset_iloc_example2 = df.iloc[10:20, 0:3]
    print(".iloc example 2 (integer-based):")
    display(subset_iloc_example2)

    # You can also select specific rows and columns by their integer positions
    # For example, the data at the 5th row (index 4) and the 2nd column (index 1)
    single_value_iloc = df.iloc[4, 1]
    print("\nValue at row 5, column 2 (iloc[4, 1]):")
    print(single_value_iloc)

except IndexError:
    print("Index out of bounds. Please check the DataFrame dimensions.")
except Exception as e:
    print(f"An error occurred: {e}")
```

# Sales Data Analysis

This notebook analyzes sales data to understand sales performance by region and product category.

## Load Data
https://realpython.com/bonus/how-to-pandas-pivot-table-code/

This cell loads the sales data from the `sales_data.csv` file into a pandas DataFrame and displays the first 5 rows to preview the data structure.

```python
import pandas as pd
df = pd.read_csv("sales_data.csv")
df.head(5)
```

## Generate Pivot Table

This cell creates a pivot table from the DataFrame to show the total `sale_price` for each `sales_region` and `product_category`. The `aggfunc='sum'` argument calculates the sum of `sale_price` for each group

```python
pivot_table = df.pivot_table(values='sale_price', index='sales_region', columns='product_category', aggfunc='sum')
display(pivot_table)
```

## Calculate Total Sales by Product Category

This cell calculates the total sales for each product category by grouping the DataFrame by `product_category` and summing the `sale_price`.

```python
total_sales_by_category = df.groupby('product_category')['sale_price'].sum()
display(total_sales_by_category)
```

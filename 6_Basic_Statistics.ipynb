#Import dataset from kaggel
https://www.kaggle.com/datasets/mosapabdelghany/medical-insurance-cost-dataset

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mosapabdelghany/medical-insurance-cost-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
df = pd.read_csv(path + '/insurance.csv')
display(df.head())
```

#Functions
.describe(), .mean(), .median(), .value_counts()

```python
df.describe()
```

.Median() Function

```python
df['charges'].median()
```

.Mean() Function

```python
df["age"].mean()
```

.value_counts() Function

```python
smoker_s = df.value_counts("smoker")
print(smoker_s)
```

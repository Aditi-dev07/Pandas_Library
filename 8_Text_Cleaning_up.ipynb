#Load dataset from kaggel Api

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("skmewati/myntra-sales-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

# Assuming the dataset is a CSV file inside the downloaded directory
# We need to find the actual file name within the downloaded path
dataset_files = os.listdir(path)
csv_file = None
for file in dataset_files:
    if file.endswith('.csv'):
        csv_file = file
        break

if csv_file:
    dataset_path = os.path.join(path, csv_file)
    df = pd.read_csv(dataset_path)
    display(df.head())
else:
    print("No CSV file found in the dataset directory.")
```

#.str accessor
Allows you to apply string methods to a Series of strings. It provides convenient ways to manipulate text data within a DataFrame or Series.

#str.upper()
This method converts all characters in each string within the Series to uppercase

```python
lw = df['brand_name'].str.lower()
print(lw.head())
```

#.str.lower()
This method converts all characters in each string within the Series to lowercase.

```python
lu = df['pants_description'].str.lower()
print(lu.head())
```

#.str.replace()
This method replaces occurrences of a specified substring with another substring in each string within the Series.`



```python
df['pants_description_replaced'] = df['pants_description'].str.replace('Jeans', 'Denim')
display(df[['pants_description', 'pants_description_replaced']].head())
```

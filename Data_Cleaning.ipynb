####**Data Cleaning - Handling Missing Values**


**Import Dataset from Kaggel**

 https://www.kaggle.com/c/titanic/data?select=test.csv


**Import Pandas as pd**

```python
import pandas as pd
```

####**Load the Dataset with pd.read_csv("") Function**

```python
df = pd.read_csv("Titanic.csv")
```

####**df.isnull to know the empty values**

```python
df.isnull()
```

** **

####**df.sum() Function to calculate total number of empty values based on columns**

```python
df.isnull().sum()
```

####**df.dropna() for drop missing values**

```python
df = df.dropna('Cabin', axis=1 , inplace=True)
```

####**df.fillna() to fill missing values by medain**

```python
median_age = df['Age'].median()
df['Age'].fillna(median_age, inplace=True)

median_fare = df['Fare'].median()
df['Fare'].fillna(median_fare, inplace=True)

print("Missing values after filling:")
print(df.isnull().sum())
```

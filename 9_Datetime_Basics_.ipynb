#Import data from Kaggel
https://www.kaggle.com/datasets/ibrohimsharipov/12-months-sales-data

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ibrohimsharipov/12-months-sales-data")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# List files in the downloaded directory
file_list = os.listdir(path)
print("Files in the dataset directory:", file_list)

# Assuming the CSV file is named "12_months_sales_data.csv" or similar,
# replace with the actual filename if different.
csv_file_name = [f for f in file_list if f.endswith('.csv')][0]
csv_file_path = os.path.join(path, csv_file_name)

# Read the CSV file into a pandas DataFrame
df = pd.read_csv(csv_file_path)

# Display the first few rows of the DataFrame
display(df.head())
```

## Data Wrangling: Convert 'Order Date' to Datetime and Extract Month and Year

The following code converts the 'Order Date' column to datetime objects and extracts the month and year into new columns for further analysis.

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
display(df.info())
```

```python
# Extract month and year using the .dt accessor
df['Order Month'] = df['Order Date'].dt.month
df['Order Year'] = df['Order Date'].dt.year

# Display the DataFrame with the new columns
display(df.head())
```

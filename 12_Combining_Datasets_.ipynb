#Import the Data from Kaggle

https://www.kaggle.com/datasets/vainero/restaurants-customers-orders-dataset?utm_source=chatgpt.com

```python
import pandas as pd
df_cities = pd.read_csv("cities.csv")
df_meals = pd.read_csv("meals.csv")
df_orders = pd.read_csv("orders.csv")
df_restaurants = pd.read_csv("restaurants.csv")
```

Display the head of the colums

```python
display(df_cities.head())
display(df_meals.head())
display(df_orders.head())
display(df_restaurants.head())
```

Display names of the colums

```python
print("Cities:", df_cities.columns)
print("Meals:", df_meals.columns)
print("Restaurants:", df_restaurants.columns)
print("Orders:", df_orders.columns)
```

#Inner Join

Returns only the rows where there is a match in both tables.

Example: Merge Orders with Restaurants on restaurant_id

```python
orders_restaurants = pd.merge(df_orders, df_restaurants, on="restaurant_id", how="inner")
print(orders_restaurants.head())
```

#Left Join

Returns all rows from the left DataFrame, and matched rows from the right. Non-matching rows will have

```python
orders_restaurants_left = pd.merge(df_orders, df_restaurants, on="restaurant_id", how="left")
print(orders_restaurants_left.head())
```

#Right Join

Returns all rows from the right DataFrame, and matched rows from the left.

```python
orders_restaurants_right = pd.merge(df_orders, df_restaurants, on="restaurant_id", how="right")
print(orders_restaurants_right.head())
```

#Outer Join

Returns all rows from both DataFrames. Non-matching rows get

```python
orders_restaurants_outer = pd.merge(df_orders, df_restaurants, on="restaurant_id", how="outer")
print(orders_restaurants_outer.head())
```

#Chaining Joins (Multiple tables)

Suppose you want to get Orders → Restaurants → Meals → Cities, you can chain merges:

```python
# Merge Orders with Restaurants
orders_restaurants = pd.merge(df_orders, df_restaurants, on="restaurant_id", how="inner")

# Merge the result with Meals
orders_restaurants_meals = pd.merge(orders_restaurants, df_meals, on="restaurant_id", how="inner")

# Merge with Cities
final_df = pd.merge(orders_restaurants_meals, df_cities, on="city_id", how="inner")

print(final_df.head())
```
